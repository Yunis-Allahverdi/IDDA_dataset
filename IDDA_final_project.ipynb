# Reading Dataset

```python
import pandas as pd

df = pd.read_csv('services_provided_to_population.csv')
df.head(10)
```

# Dataset Cleaning and Processing

- Dropping irrelevant columns, rows and resetting the index
- Renaming columns
- Converting data types to `float64` for numerical analysis
- Dataset understanding (`df.describe`, `df.info`)
- Feature Engineering: Creating a `Total` volume and an `Education_total` aggregate.

```python
df = df.drop(columns='Unnamed: 1')
df = df.drop(0).reset_index(drop=True)
```

```python
for col in df.columns:
    if col != 'Year':
        df[col] = df[col].astype('float64')

df.info()
```

```python
df.describe().round(2)
```

```python
education_cols = ['Education_General', 'Education_Preschool', 'Education_Higher_Secondary', 'Education_Courses']
df['Education_total'] = df[education_cols].sum(axis=1)
sector_cols = [
    'Household', 'Transportation', 'Communication', 'Housing',
    'Utilities', 'Culture', 'Tourism', 'Sports',
    'Healthcare', 'Health', 'Legal_Banking', 'Education_total',
    'Other'
]
df['Total'] = df[sector_cols].sum(axis=1)
df.head()
```

### Pre vs. Post COVID-19 Analysis

```python
df = df.sort_values('Year')
df_pre = df[df['Year'] < 2020]
df_post = df[df['Year'] >= 2020]
pre_growth = df_pre[sector_cols].pct_change().median() * 100
pre_growth = pre_growth.sort_values(ascending=False)
post_growth = df_post[sector_cols].pct_change().median() * 100
post_growth = post_growth.sort_values(ascending=False)
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Creating the layout
graph = make_subplots(
    rows=2, cols=1, 
    subplot_titles=('Median Annual Growth Rate by Sector (Before COVID)', 
                    'Median Annual Growth Rate by Sector (After COVID)')
)

# Adding the bars to the layout
graph.add_trace(go.Bar(x=pre_growth.index, y=pre_growth.values), row=1, col=1)
graph.add_trace(go.Bar(x=post_growth.index, y=post_growth.values), row=2, col=1)

# Formatting the table by setting dimensions, and also fixed x and y axis range
graph.update_layout(
    height=1000, 
    width=1200, 
    showlegend=False,
    # This locks the top chart
    xaxis_fixedrange=True, 
    yaxis_fixedrange=True,
    # This locks the bottom chart
    xaxis2_fixedrange=True, 
    yaxis2_fixedrange=True
)

# I set the Y-axis range and format for both subplots for comparison
graph.update_yaxes(range=[0, 50], tickformat=".2f", ticksuffix="%", row=1, col=1)
graph.update_yaxes(range=[0, 50], tickformat=".2f", ticksuffix="%", row=2, col=1)

graph.show()
```

```python
# Calculating the percentage change between 2019 and 2020
df_covid = df[df['Year'].isin([2019, 2020])].sort_values('Year')
covid_impact = df_covid[sector_cols].pct_change() * 100
# Taking the results and sorting them to see which sectors were hit hardest
covid_impact = covid_impact.iloc[1].sort_values()
# Creating the bar chart to visualize impact rate on each sector
fig = px.bar(x=covid_impact.index, 
             y=covid_impact.values,
             title='COVID Impact on Sectors',
             labels={'x':'Sectors', 'y':'Impact Rate (%)'},
             width=1500, 
             height=700)
# Adding a red line at 0 to show the baseline clearly
fig.add_hline(y=0, line_color="red")
# Setting the y-axis range to show the drop and fixed the axis to keep it stable
fig.update_layout(yaxis=dict(range=[5, -100], tickformat=".3s", ticksuffix="%"))
fig.update_layout(xaxis_fixedrange=True, yaxis_fixedrange=True)
fig.show()
```

```python
yoy_growth = df[sector_cols].pct_change() * 100  # Year-over-Year growth percentage for all sectors
corr_matrix = yoy_growth.corr() # Creating the corr matrix
corr_matrix = corr_matrix.round(2)  # Rounding decimal numbers to make it easier to read
```

```python
# Creating the heatmap to visualize
graph = px.imshow(
    corr_matrix, 
    text_auto='.2f', 
    color_continuous_scale='RdBu_r', 
    title='Correlation Between Sector Growth Rates',
    width=1000, 
    height=900)
# Applying fixed range and a custom hovertemplate
graph.update_layout(xaxis_fixedrange=True, yaxis_fixedrange=True)
graph.update_traces(hovertemplate="Sector X: %{x}<br>Sector Y: %{y}<br>Correlation: %{z}<extra></extra>")
graph.show()
```

### 10-Year Growth Forecasting (Using Log-Linear Regression Model)

```python
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.linear_model import LinearRegression

# 1. Preparing Data for next 10 years starting from 2026
X = df[['Year']].values
future_years = np.array(range(2026, 2036)).reshape(-1, 1)
forecast_results = {'Year': future_years.flatten()}

# 2. Simple Loop for Predictions
for sector in sector_cols:
    y_log = np.log1p(df[sector].values)
    model = LinearRegression().fit(X, y_log)
    forecast_results[sector] = np.expm1(model.predict(future_years))

forecast_df = pd.DataFrame(forecast_results)
```

```python
# Creating the line graph to visualize it
graph = px.line(forecast_df, 
              x='Year', 
              y=sector_cols,
              title='10-Year Service Sector Volume Forecast (Log-Linear Regression)',
              labels={'value':'Volume', 'Year':'Year'},
              markers=True,
              width=1500, 
              height=900)

# Applying fixed layout and custom hovertemplate
graph.update_layout(
    yaxis_tickformat='.3s', 
    xaxis_fixedrange=True,  
    yaxis_fixedrange=True,   
    legend_title_text='Sectors')

graph.update_traces(hovertemplate="Sector: %{fullData.name}<br>Year: %{x}<br>Volume: %{y}<extra></extra>")
graph.show()
```
